## 🔮 TEST 2: Query Cost Predictor

````python
from airt.compiler.query_predictor import QueryCostPredictor

predictor = QueryCostPredictor()
queries = [
    "Hello",
    "What is the capital of France?",
    "Explain quantum computing in simple terms",
    "Write a Python function to implement a neural network",
    "Prove the Riemann hypothesis",
]

print(f"{'🔮 Query Cost Predictor':^60}")
print(f"{'─'*60}")
for q in queries:
    r = predictor.predict_cost(q)
    cost = r['cost_score']
    meter = '█' * int(cost * 20) + '░' * (20 - int(cost * 20))
    print(f"  [{meter}] {r['category']:8s} | {r['recommended_precision']:4s} | {q[:40]}")
print(f"{'─'*60}")
print("✅ Query Cost Predictor works!")
print("\n💡 This predicts compute cost BEFORE running the model.")
````

## 🖥️ TEST 4: Start Web Server

````python
# Start the FastAPI server in background
import subprocess, threading, time

def run_server():
    subprocess.run(["python", "server/api.py"], capture_output=False)

server_thread = threading.Thread(target=run_server, daemon=True)
server_thread.start()
time.sleep(2)

# Check if server is running
import requests
try:
    r = requests.get("http://localhost:8000/api/v1/info", timeout=5)
    print("✅ Server is running!")
    print(r.json())
except:
    print("⚠️ Server may need port forwarding in Colab")
    print("Use: from google.colab.output import eval_js")
    print("     print(eval_js(\"google.colab.kernel.proxyPort(8000)\"))")
````
