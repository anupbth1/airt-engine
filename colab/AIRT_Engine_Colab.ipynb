## 🔮 STEP 3: Query Cost Predictor (Novel Feature)
Predicts compute cost BEFORE running the model.
---

````python
from airt.compiler.query_predictor import QueryCostPredictor

predictor = QueryCostPredictor()
queries = [
    "Hello",
    "What is the capital of France?",
    "Explain quantum computing in simple terms",
    "Write a Python function to implement a neural network",
    "Prove the Riemann hypothesis",
]

print(f"{'='*60}")
print(f"{'🔮 Query Cost Predictor':^60}")
print(f"{'='*60}")
print()
for q in queries:
    r = predictor.predict_cost(q)
    cost = r['cost_score']
    meter = '█' * int(cost * 20) + '░' * (20 - int(cost * 20))
    print(f"  [{meter}] {r['category']:8s} | {r['recommended_precision']:4s} | Cost: {cost:.0%}")
    print(f"          Tokens: ~{r['estimated_tokens']} | {q}")
    print()
print(f"{'='*60}")
print("✅ Query Cost Predictor works without GPU!")
````

````python
# Start the server in background
import subprocess, threading, time

def run_server():
    subprocess.run(["python", "cli.py", "server"], capture_output=False)

server_thread = threading.Thread(target=run_server, daemon=True)
server_thread.start()
time.sleep(3)

# Get Colab proxy URL
try:
    from google.colab.output import eval_js
    url = eval_js("google.colab.kernel.proxyPort(8000)")
    print(f"🌐 Open Web UI: {url}")
    print(f"🌐 API Info: {url}api/v1/info")
except ImportError:
    print("⚠️ Not running in Colab. Server at http://localhost:8000")
````
